```python
import sqlite3
import pandas as pd
from pathlib import Path

# 1. Connect to your database
db_path = Path('../data/cve_database.db')
conn = sqlite3.connect(db_path)

# 2. Read the enrichments table into a Pandas DataFrame
query = "SELECT * FROM enrichments where chpl_flag != 0 LIMIT 10"
df = pd.read_sql_query(query, conn)

query1 = "SELECT count(*) FROM enrichments where epss_score IS NOT NULL"
x = pd.read_sql_query(query1, conn)
print(x);
```
