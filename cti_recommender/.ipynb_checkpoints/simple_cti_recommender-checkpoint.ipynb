```python
import requests
import pandas as pd
from datetime import datetime, timezone
from dateutil.parser import parse as parse_date
```

```python
# -------------------------
# CONFIG
# -------------------------

# NVD API: we will pull recent CVEs (e.g., last 7 days) as a starting point.
# Later you can expand to sector filters.
NVD_API_URL = "https://services.nvd.nist.gov/rest/json/cves/2.0"

# CISA KEV catalog (CSV)
CISA_KEV_URL = "https://www.cisa.gov/sites/default/files/csv/known_exploited_vulnerabilities.csv"
```

```python
def fetch_nvd_recent_cves(days_back: int = 7) -> pd.DataFrame:
    """
    Fetch recent CVEs from NVD last `days_back` days.
    Very first simple pull.
    """
    print(f"[*] Fetching NVD CVEs from last {days_back} days...")

    now = datetime.now(timezone.utc)
    start = now.replace(hour=0, minute=0, second=0, microsecond=0) - pd.Timedelta(days=days_back)
    pub_start = start.isoformat(timespec="seconds").replace("+00:00", "Z")
    pub_end = now.isoformat(timespec="seconds").replace("+00:00", "Z")

    params = {
        "pubStartDate": pub_start,
        "pubEndDate": pub_end,
        "resultsPerPage": 2000,  # max allowed; adjust as needed
    }

    resp = requests.get(NVD_API_URL, params=params, timeout=60)
    resp.raise_for_status()
    data = resp.json()

    records = []
    for item in data.get("vulnerabilities", []):
        cve = item.get("cve", {})
        cve_id = cve.get("id")
        published = cve.get("published")
        descriptions = cve.get("descriptions", [])
        desc = ""
        if descriptions:
            desc = descriptions[0].get("value", "")

        # CVSS v3.1 if available
        metrics = cve.get("metrics", {})
        cvss_score = None

        # Try CVSS v3.1
        cvss_v31 = metrics.get("cvssMetricV31", [])
        if cvss_v31:
            cvss_score = cvss_v31[0]["cvssData"]["baseScore"]
        else:
            # fallback v3.0
            cvss_v3 = metrics.get("cvssMetricV30", [])
            if cvss_v3:
                cvss_score = cvss_v3[0]["cvssData"]["baseScore"]

        records.append({
            "cve_id": cve_id,
            "published": published,
            "description": desc,
            "cvss": cvss_score
        })

    df = pd.DataFrame(records)
    print(f"[*] Fetched {len(df)} CVEs from NVD")
    return df


def fetch_cisa_kev() -> pd.DataFrame:
    """
    Fetch CISA KEV catalog as DataFrame.
    """
    print("[*] Fetching CISA KEV catalog...")
    resp = requests.get(CISA_KEV_URL, timeout=60)
    resp.raise_for_status()

    # Read CSV from content
    from io import StringIO
    kev_df = pd.read_csv(StringIO(resp.text))

    # Standardize column names
    kev_df.columns = [c.strip().lower().replace(" ", "_") for c in kev_df.columns]
    kev_df.rename(columns={"cve_id": "cve_id"}, inplace=True)

    print(f"[*] Fetched {len(kev_df)} KEV entries")
    return kev_df


def compute_recency_score(published: str, now: datetime) -> float:
    """
    Simple recency score: newer CVEs get higher score.
    0-1 scale (clipped).
    """
    if not isinstance(published, str):
        return 0.0

    try:
        pub_date = parse_date(published)
    except Exception:
        return 0.0

    days_diff = (now - pub_date).days
    # example: <= 7 days -> 1.0; 180+ days -> ~0.0
    max_days = 180
    score = max(0.0, 1.0 - (days_diff / max_days))
    return float(round(score, 3))


def build_simple_score(df: pd.DataFrame) -> pd.DataFrame:
    """
    Build a simple AI-ish score using:
    - recency
    - KEV flag
    - CVSS
    This is your first prototype scoring engine.
    """
    print("[*] Building simple scores...")

    now = datetime.now(timezone.utc)
    df["recency_score"] = df["published"].apply(lambda x: compute_recency_score(x, now))

    # Normalise CVSS to 0-1
    max_cvss = 10.0
    df["cvss_norm"] = df["cvss"].fillna(0) / max_cvss

    # KEV flag already 0/1
    # Weighting (you can tune these)
    w_recency = 0.4
    w_kev = 0.4
    w_cvss = 0.2

    df["final_score"] = (
        w_recency * df["recency_score"] +
        w_kev * df["kev_flag"] +
        w_cvss * df["cvss_norm"]
    )

    return df


def main():
    # Step 1: Fetch data
    nvd_df = fetch_nvd_recent_cves(days_back=30)
    kev_df = fetch_cisa_kev()

    # Step 2: Merge NVD with KEV
    kev_df_small = kev_df[["cve_id"]].dropna().drop_duplicates()
    kev_df_small["kev_flag"] = 1

    merged = pd.merge(
        nvd_df,
        kev_df_small,
        on="cve_id",
        how="left"
    )

    merged["kev_flag"] = merged["kev_flag"].fillna(0).astype(int)

    # Step 3: Score
    scored = build_simple_score(merged)

    # Step 4: Sort by score descending
    scored_sorted = scored.sort_values(by="final_score", ascending=False)

    # Step 5: Show top 20 as first recommender output
    print("\n==== Top 20 Recommended CVEs (very first prototype) ====\n")
    cols_to_show = ["cve_id", "published", "cvss", "kev_flag", "recency_score", "final_score", "description"]
    print(scored_sorted[cols_to_show].head(20).to_string(index=False))


if __name__ == "__main__":
    main()

# -------------------------
# CONFIG
# -------------------------

# NVD API: we will pull recent CVEs (e.g., last 7 days) as a starting point.
# Later you can expand to sector filters.
NVD_API_URL = "https://services.nvd.nist.gov/rest/json/cves/2.0"

# CISA KEV catalog (CSV)
CISA_KEV_URL = "https://www.cisa.gov/sites/default/files/csv/known_exploited_vulnerabilities.csv"


def fetch_nvd_recent_cves(days_back: int = 7) -> pd.DataFrame:
    """
    Fetch recent CVEs from NVD last `days_back` days.
    Very first simple pull.
    """
    print(f"[*] Fetching NVD CVEs from last {days_back} days...")

    now = datetime.now(timezone.utc)
    start = now.replace(hour=0, minute=0, second=0, microsecond=0) - pd.Timedelta(days=days_back)
    pub_start = start.isoformat(timespec="seconds").replace("+00:00", "Z")
    pub_end = now.isoformat(timespec="seconds").replace("+00:00", "Z")

    params = {
        "pubStartDate": pub_start,
        "pubEndDate": pub_end,
        "resultsPerPage": 2000,  # max allowed; adjust as needed
    }

    resp = requests.get(NVD_API_URL, params=params, timeout=60)
    resp.raise_for_status()
    data = resp.json()

    records = []
    for item in data.get("vulnerabilities", []):
        cve = item.get("cve", {})
        cve_id = cve.get("id")
        published = cve.get("published")
        descriptions = cve.get("descriptions", [])
        desc = ""
        if descriptions:
            desc = descriptions[0].get("value", "")

        # CVSS v3.1 if available
        metrics = cve.get("metrics", {})
        cvss_score = None

        # Try CVSS v3.1
        cvss_v31 = metrics.get("cvssMetricV31", [])
        if cvss_v31:
            cvss_score = cvss_v31[0]["cvssData"]["baseScore"]
        else:
            # fallback v3.0
            cvss_v3 = metrics.get("cvssMetricV30", [])
            if cvss_v3:
                cvss_score = cvss_v3[0]["cvssData"]["baseScore"]

        records.append({
            "cve_id": cve_id,
            "published": published,
            "description": desc,
            "cvss": cvss_score
        })

    df = pd.DataFrame(records)
    print(f"[*] Fetched {len(df)} CVEs from NVD")
    return df


def fetch_cisa_kev() -> pd.DataFrame:
    """
    Fetch CISA KEV catalog as DataFrame.
    """
    print("[*] Fetching CISA KEV catalog...")
    resp = requests.get(CISA_KEV_URL, timeout=60)
    resp.raise_for_status()

    # Read CSV from content
    from io import StringIO
    kev_df = pd.read_csv(StringIO(resp.text))

    # Standardize column names
    kev_df.columns = [c.strip().lower().replace(" ", "_") for c in kev_df.columns]
    kev_df.rename(columns={"cve_id": "cve_id"}, inplace=True)

    print(f"[*] Fetched {len(kev_df)} KEV entries")
    return kev_df


def compute_recency_score(published: str, now: datetime) -> float:
    """
    Simple recency score: newer CVEs get higher score.
    0-1 scale (clipped).
    """
    if not isinstance(published, str):
        return 0.0

    try:
        pub_date = parse_date(published)
    except Exception:
        return 0.0

    days_diff = (now - pub_date).days
    # example: <= 7 days -> 1.0; 180+ days -> ~0.0
    max_days = 180
    score = max(0.0, 1.0 - (days_diff / max_days))
    return float(round(score, 3))


def build_simple_score(df: pd.DataFrame) -> pd.DataFrame:
    """
    Build a simple AI-ish score using:
    - recency
    - KEV flag
    - CVSS
    This is your first prototype scoring engine.
    """
    print("[*] Building simple scores...")

    now = datetime.now(timezone.utc)
    df["recency_score"] = df["published"].apply(lambda x: compute_recency_score(x, now))

    # Normalise CVSS to 0-1
    max_cvss = 10.0
    df["cvss_norm"] = df["cvss"].fillna(0) / max_cvss

    # KEV flag already 0/1
    # Weighting (you can tune these)
    w_recency = 0.4
    w_kev = 0.4
    w_cvss = 0.2

    df["final_score"] = (
        w_recency * df["recency_score"] +
        w_kev * df["kev_flag"] +
        w_cvss * df["cvss_norm"]
    )

    return df


def main():
    # Step 1: Fetch data
    nvd_df = fetch_nvd_recent_cves(days_back=30)
    kev_df = fetch_cisa_kev()

    # Step 2: Merge NVD with KEV
    kev_df_small = kev_df[["cve_id"]].dropna().drop_duplicates()
    kev_df_small["kev_flag"] = 1

    merged = pd.merge(
        nvd_df,
        kev_df_small,
        on="cve_id",
        how="left"
    )

    merged["kev_flag"] = merged["kev_flag"].fillna(0).astype(int)

    # Step 3: Score
    scored = build_simple_score(merged)

    # Step 4: Sort by score descending
    scored_sorted = scored.sort_values(by="final_score", ascending=False)

    # Step 5: Show top 20 as first recommender output
    print("\n==== Top 20 Recommended CVEs (very first prototype) ====\n")
    cols_to_show = ["cve_id", "published", "cvss", "kev_flag", "recency_score", "final_score", "description"]
    print(scored_sorted[cols_to_show].head(20).to_string(index=False))


if __name__ == "__main__":
    main()
```
